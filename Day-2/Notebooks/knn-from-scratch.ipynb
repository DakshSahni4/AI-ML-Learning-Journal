```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/kaggle/input/datasets/dakshsahni567/cleaned-iris/cleaned_dirty_iris.csv")
```

```python
X = df.drop("species",axis = 1)
y = df["species"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state = 2026,test_size = 0.2,stratify = y)
```

```python
def KNNClassifier(X_train,X_test):
    k = 2
    pred = pd.DataFrame(columns =["pred"])
    for row in range(len(X_test)):
        distances = np.sqrt(((X_train-X_test.iloc[row])**2).sum(axis = 1))
        nearest_indices = distances.nsmallest(k).index.tolist()

        nearest_labels = y_train.loc[nearest_indices]

        prediction = nearest_labels.mode()[0]
        pred.loc[len(pred)] = prediction
    return pred
        
```

```python
pred1 = KNNClassifier(X_train,X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,pred1)
```
